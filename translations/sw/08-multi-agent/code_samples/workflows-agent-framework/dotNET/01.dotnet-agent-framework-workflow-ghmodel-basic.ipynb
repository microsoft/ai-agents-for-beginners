# 🔄 Mipangilio ya Msingi ya Wakala kwa kutumia Azure OpenAI (API za Majibu) (.NET)

## 📋 Mafunzo ya Mipangilio ya Kazi

Daftari hili linaonyesha jinsi ya kujenga **miradi ya wakala tata** kwa kutumia Microsoft Agent Framework kwa .NET na Azure OpenAI (API za Majibu). Utajifunza kuunda mchakato wa biashara wa hatua nyingi ambapo mawakala wa AI hushirikiana kutimiza kazi ngumu kupitia mifumo ya mpangilio iliyopangwa.

## 🎯 Malengo ya Kujifunza

### 🏗️ **Misingi ya Miundo ya Mipangilio ya Kazi**
- **Mjengaji wa Mipangilio**: Tengeneza na panga michakato tata ya AI yenye hatua nyingi
- **Urithishaji Mdhibiti**: Ratibu mawakala wengi maalum ndani ya mipangilio
- **Azure OpenAI (API za Majibu)**: Tumia Azure OpenAI Responses API katika mipangilio
- **Ubunifu wa Mipangilio kwa Macho**: Unda na ona muundo wa mipangilio kwa uelewa bora

### 🔄 **Mifumo ya Mpangilio wa Mchakato**
- **Usindikaji wa Mfuatano**: Unganisha kazi nyingi za wakala kwa mpangilio wa mantiki
- **Usimamizi wa Hali**: Dumisha muktadha na mtiririko wa data kupitia hatua za mipangilio
- **Ushughulikiaji wa Makosa**: Tekeleza urejeshaji thabiti wa makosa na ustahimilivu wa mipangilio
- **Uboreshaji wa Utendaji**: Tengeneza mipangilio yenye ufanisi kwa shughuli za kiwango cha shirika

### 🏢 **Matumizi ya Mipangilio ya Shirika**
- **Mifumo ya Kuendesha Biashara**: Tengeneza otomatiki mipangilio tata ya shirika
- **Mtiririko wa Uzalishaji wa Maudhui**: Mipangilio ya uhariri yenye hatua za mapitio na idhini
- **Otomatiki ya Huduma kwa Wateja**: Utatuzi wa maswali ya wateja yenye hatua nyingi
- **Mipangilio ya Usindikaji Data**: Mipangilio ya ETL yenye mabadiliko yanayotolewa na AI

### 🔑 **Usanidi wa Azure OpenAI**

**Usanidi wa Mazingira (faili la .env):**
```env
AZURE_OPENAI_ENDPOINT=https://<your-resource>.openai.azure.com
AZURE_OPENAI_DEPLOYMENT=gpt-4o-mini
```

**Upatikanaji wa Azure OpenAI:**
1. Unda rasilimali ya Azure OpenAI kwenye dirisha la Azure
2. Tengeneza mfano (kwa mfano, `gpt-4o-mini`) na kumbuka jina la usambazaji
3. Ingia kwa kutumia `az login` na usanidi vigezo vya mazingira kama ilivyoelezwa hapo juu

````polyglot-notebook
using System;
using System.ComponentModel;
using Azure.AI.OpenAI;
using Azure.Identity;
using Microsoft.Extensions.AI;
using Microsoft.Agents.AI;
using Microsoft.Agents.AI.Workflows;
````

````polyglot-notebook
 using DotNetEnv;
````

````polyglot-notebook
Env.Load("../../../.env");
````

````polyglot-notebook
// Azure OpenAI with the Responses API (stable v1 endpoint). Sign in with `az login`.
var azureEndpoint = Environment.GetEnvironmentVariable("AZURE_OPENAI_ENDPOINT") ?? throw new InvalidOperationException("AZURE_OPENAI_ENDPOINT is not set.");
var deployment = Environment.GetEnvironmentVariable("AZURE_OPENAI_DEPLOYMENT") ?? "gpt-4o-mini";
````

````polyglot-notebook
// The Azure OpenAI client is created directly from the endpoint and Azure CLI credential — no custom client options are required.
````

````polyglot-notebook
var azureClient = new AzureOpenAIClient(new Uri(azureEndpoint), new AzureCliCredential());
````

````polyglot-notebook
const string ReviewerAgentName = "Concierge";
const string ReviewerAgentInstructions = @"
    You are an are hotel concierge who has opinions about providing the most local and authentic experiences for travelers.
    The goal is to determine if the front desk travel agent has recommended the best non-touristy experience for a traveler.
    If so, state that it is approved.
    If not, provide insight on how to refine the recommendation without using a specific example. ";
````

````polyglot-notebook
const string FrontDeskAgentName = "FrontDesk";
const string FrontDeskAgentInstructions = @"""
    You are a Front Desk Travel Agent with ten years of experience and are known for brevity as you deal with many customers.
    The goal is to provide the best activities and locations for a traveler to visit.
    Only provide a single recommendation per response.
    You're laser focused on the goal at hand.
    Don't waste time with chit chat.
    Consider suggestions when refining an idea.
    """;
````

````polyglot-notebook
AIAgent reviewerAgent = azureClient.GetOpenAIResponseClient(deployment).CreateAIAgent(
    name:ReviewerAgentName,instructions:ReviewerAgentInstructions);
AIAgent frontDeskAgent  = azureClient.GetOpenAIResponseClient(deployment).CreateAIAgent(
    name:FrontDeskAgentName,instructions:FrontDeskAgentInstructions);
````

````polyglot-notebook
var workflow = new WorkflowBuilder(frontDeskAgent)
            .AddEdge(frontDeskAgent, reviewerAgent)
            .Build();
````

````polyglot-notebook
ChatMessage userMessage = new ChatMessage(ChatRole.User, [
	new TextContent("I would like to go to Paris.") 
]);
````

````polyglot-notebook
StreamingRun run = await InProcessExecution.StreamAsync(workflow, userMessage);
````

````polyglot-notebook
await run.TrySendMessageAsync(new TurnToken(emitEvents: true));
string id="";
string messageData="";
await foreach (WorkflowEvent evt in run.WatchStreamAsync().ConfigureAwait(false))
{
    if (evt is AgentRunUpdateEvent executorComplete)
    {
        if(id=="")
        {
            id=executorComplete.ExecutorId;
        }
        if(id==executorComplete.ExecutorId)
        {
            messageData+=executorComplete.Data.ToString();
        }
        else
        {
            id=executorComplete.ExecutorId;
        }
        // Console.WriteLine($"{executorComplete.ExecutorId}: {executorComplete.Data}");
    }
}

Console.WriteLine(messageData);
````
